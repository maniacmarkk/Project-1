```python
def get_student_data():
    student = {}
    name = input("Enter student name: ")
    score = input("Enter each score separated by a space: ")
    while score.strip() == "":
        score = input("Enter at least one score: ")
    l = score.split()
    intList = [int(i) for i in l]
    student["Name"] = name
    student["Scores"] = intList
    avg = sum(student["Scores"]) / len(student["Scores"])
    student["Average"] = avg
    student["Highest"] = max(student["Scores"])
    student["Lowest"] = min(student["Scores"])

    if avg >= 90:
        student["Grade"] = "A"
    elif avg >= 80:
        student["Grade"] = "B"
    elif avg >= 70:
        student["Grade"] = "C"
    elif avg >= 60:
        student["Grade"] = "D"
    else:
        student["Grade"] = "F"
    
    return student

def print_student_summary(student):
    for key, value in student.items():
        print(f"{key}: {value}")



def main():
    students = []
    num = int(input("How many students? "))
    while num <= 0:
        num = int(input("Enter at least one student: "))
    for i in range(num):
        student = get_student_data()
        students.append(student)
    cumulative = 0
    best_score_index = 0
    best_avg = students[0]["Average"]
    for i in range(1, num):
        if (students[i]["Average"] > students[best_score_index]["Average"]):
            best_score_index = i
            best_avg = students[best_score_index]["Average"]
    for student in students:
        print_student_summary(student)
        print()
        cumulative += student["Average"]
    print("Class average:", cumulative/num)
    print("Top student: " + students[best_score_index]["Name"])
    print("Average:", students[best_score_index]["Average"])
    answer = False
    ans = input("Do you want to find a student's information? ").strip().lower()
    if ans in ("true", "t", "yes", "y", "1"):
        answer = True
    if answer: 
        toFind = input("Which student do you want to find? ")
        print()
        found = False
        for s in students:
            if s["Name"] == toFind:
                print_student_summary(s)
                found = True
            
        if not found:
            print("We don't have record of " + toFind)
    
main()




    

        
        
```
